```python
import pandas as pd 
from pathlib import Path 
```

```python
file_path = Path("C:\\Users\\Anna May\\Downloads\\FANRBF\\TVET_Stats_Analysis\\tvet_statistics_analysis\\data\\raw\\raw_tvet_data.xls")

excel = pd.ExcelFile(file_path)
print(excel.sheet_names)
```

```python
raw = pd.read_excel(file_path, header=None)
display(raw.head(12))
```

```python
columns = [
    "region",
    "province",
    "assessed_female",
    "assessed_male",
    "assessed_total",
    "certified_female",
    "certified_male",
    "certified_total",
    "certification_rate"
]

df = raw.iloc[5:, :9].copy()
df.columns = columns

display(df.head())
```

```python
#text formatting for region and province columns
for col in ["region", "province"]:
    df[col] = df[col].astype(str).str.strip()

print(df.head())
```

```python
#dropping rows
df = df.dropna(how="all")
df = df[df['province'].notna() & df['province'].ne("nan")].copy()

print(df)
```

```python
#filling down the region column with the appropriate values
df['region'] = df['region'].replace("nan", pd.NA).ffill()

print(df.head(10))
```

```python
print(df.dtypes)
```

```python
#change data types for numeric columns
numeric_columns = [
    "assessed_female",
    "assessed_male",
    "assessed_total",
    "certified_female",
    "certified_male",
    "certified_total"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce").astype("Int64")

df["certification_rate"] = pd.to_numeric(df["certification_rate"], errors="coerce")
df[["region", "province"]] = df[["region", "province"]].astype("str")

print(df.dtypes)
```

```python
#check for missing values in the dataset
print(df.isna().sum())
```

```python
print(df[df["assessed_female"].isna()])
```

```python
print(type(df))
print(type(numeric_columns))
```

```python
#replace the missing values in the numeric columns with 0
df[numeric_columns] = df[numeric_columns].fillna(0)
```

## Data Cleaning Log — January 2021 TVET Assessment and Certification

**Source structure:** The source is a formatted statistical report with multi-row headers, region labels that are not repeated for each province, and regional subtotal rows.

**1 — Header normalization:** Replaced the multi-level report headers with unique column names for region, province, assessed counts by sex, certified counts by sex, and certification rate.

**2 — Region labels:** Filled down the region values so that each province-level record contains its corresponding region.

**3 — Regional subtotals:** Excluded rows labeled as regional totals from the province-level analytical table. These totals will be retained separately for reconciliation where appropriate.

**4 — Numeric conversion:** Converted count columns to integers where complete and valid, and converted certification rates to numeric percentage-point values. Investigated conversion failures before resolving them.

**5 — Zero and missing values:** Investigated zero counts and incomplete sex-specific values against the original source. Retained rows with missing values as zero counts.

**Limitations:** The report covers January 1–31, 2021. The data is a snapshot for a single reporting period and cannot, on its own, establish long-term trends.
